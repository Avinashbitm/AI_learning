```python
from transformers import pipeline

classifier = pipeline(task = "sentiment-analysis")

text = "The new iphone is released with better camera and improved battery life"

results = classifier(text)

print(results)
```

```python
from transformers import pipeline

classifier = pipeline(task = "zero-shot-classification",model ="facebook/bart-large" )

text = "The new iphone is released with better camera and improved battery life"

candidate_labels = ["Technology","politics","sports","health"]

results = classifier(text,candidate_labels)

print(results)
```

```python
from torch.nn.modules import padding
from transformers import pipeline

classifier = pipeline("sentiment-analysis")

text = ["This file was a waste of time","This file was very nice"]
tokenizer = classifier.tokenizer

results = tokenizer(text,padding=True,return_tensors="pt")

print(results)
```

```python
model = classifier.model

output = model(
    results['input_ids'],
    results['attention_mask']
)

print(output)
```

```python
import torch

prob_dist = torch.nn.functional.softmax(output.logits,dim=1)

print(prob_dist)
```
